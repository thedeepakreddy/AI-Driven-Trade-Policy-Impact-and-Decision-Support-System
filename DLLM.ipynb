```python
from unsloth import FastVisionModel
import torch

# 1. Load the Base Model in 4-bit
model, tokenizer = FastVisionModel.from_pretrained(
    model_name = "unsloth/Qwen2.5-VL-7B-Instruct",
    load_in_4bit = True, # CRITICAL: This compresses it to fit on the Colab T4 GPU
    use_gradient_checkpointing = "unsloth",
)

# 2. Configure the Training Adapters (LoRA)
model = FastVisionModel.get_peft_model(
    model,
    finetune_vision_layers = True,     # Trains the "eyes" to read your specific UI/screenshots
    finetune_language_layers = True,   # Trains the "brain" to output your exact JSON schemas
    finetune_attention_modules = True,
    finetune_mlp_modules = True,

    r = 16,            # The size of the adapter. 16 is the sweet spot for multimodal tasks.
    lora_alpha = 16,
    lora_dropout = 0,
    bias = "none",
    random_state = 3407,
    use_rslora = False,
    loftq_config = None,
)

print("Model successfully loaded and prepped for fine-tuning!")
```

```python
from datasets import load_dataset

# Load a sample dataset. For agentic desktop tasks, you would eventually use
# your own dataset of screenshots and actions.
# Here, we use a placeholder or a common VQA dataset to demonstrate the flow.
dataset = load_dataset("unsloth/Radiology_mini", split = "train")

print(f"Dataset loaded with {len(dataset)} samples.")
display(dataset[0])
```

### Formatting the Dataset
Unsloth's Qwen2.5-VL implementation expects a specific format. We need to map our dataset columns to the conversation format (User/Assistant) that includes images.

```python
from unsloth import standardize_sharegpt
from datasets import concatenate_datasets

# 1. Inspect dataset to find correct keys
vision_keys = dataset.column_names
math_keys = math_dataset.column_names

# Dynamic key detection based on your actual kernel state:
# Vision keys: ['image', 'image_id', 'caption', 'cui']
vision_text_key = "caption" if "caption" in vision_keys else "instruction"
vision_label_key = "caption" # In this mini-dataset, the caption is the label

def format_vision_data(examples):
    conversations = []
    for i in range(len(examples[vision_text_key])):
        convo = [
            {
                "role": "user",
                "content": [
                    {"type": "image", "image": examples["image"][i]},
                    {"type": "text", "text": "Describe this image."},
                ],
            },
            {
                "role": "assistant",
                "content": [
                    {"type": "text", "text": examples[vision_text_key][i]},
                ],
            },
        ]
        conversations.append(convo)
    return {"conversations": conversations}

dataset = dataset.map(format_vision_data, batched = True, remove_columns = vision_keys)
print(f"Vision dataset formatted using key: {vision_text_key}")

# 2. Format the Math dataset
def format_math_data(examples):
    conversations = []
    for i in range(len(examples["problem"])):
        convo = [
            {
                "role": "user",
                "content": [{"type": "text", "text": examples["problem"][i]}],
            },
            {
                "role": "assistant",
                "content": [{"type": "text", "text": examples["solution"][i]}],
            },
        ]
        conversations.append(convo)
    return {"conversations": conversations}

math_dataset = math_dataset.map(format_math_data, batched = True, remove_columns = math_keys)
print("Math dataset formatted!")

# 3. Combine for Multi-task training
combined_dataset = concatenate_datasets([dataset, math_dataset.select(range(min(2000, len(math_dataset))))])
combined_dataset = combined_dataset.shuffle(seed = 3407)
print(f"Combined dataset ready with {len(combined_dataset)} samples!")
```

```python
from trl import SFTTrainer
from transformers import TrainingArguments
from unsloth import is_bfloat16_supported

trainer = SFTTrainer(
    model = model,
    tokenizer = tokenizer,
    train_dataset = combined_dataset,
    dataset_text_field = "conversations",
    max_seq_length = 2048,
    dataset_num_proc = 2,
    args = TrainingArguments(
        per_device_train_batch_size = 2,
        gradient_accumulation_steps = 4,
        warmup_steps = 5,
        max_steps = 60, # Set to a low number for the first run to test
        learning_rate = 2e-4,
        fp16 = not is_bfloat16_supported(),
        bf16 = is_bfloat16_supported(),
        logging_steps = 1,
        optim = "adamw_8bit",
        weight_decay = 0.01,
        lr_scheduler_type = "linear",
        seed = 3407,
        output_dir = "outputs",
    ),
)

# Start the training
trainer_stats = trainer.train()
print("Training complete!")
```
